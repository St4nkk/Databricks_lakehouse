# Read from CSV file and write to table in Bronze Layer

```python
for file in CONFIG_INFO:
    df = (
        spark.read.option("header", "true")
        .option("inferSchema", "true")
        .csv(file["path"])
    )
    df.write.mode("overwrite").saveAsTable(f"workspace.bronze.{file["table"]}")
    print(f"File {file['path']} ingested into table {file['table']}")
```

# Check weather loaded correctly

```python
for file in CONFIG_INFO:
    df = spark.read.table(f"workspace.bronze.{file['table']}")
    print(f"Table {file['table']} contains {df.count()} records")
    display(df.limit(10))
    
```
